```python
import joblib
import matplotlib.pyplot as plt
import numpy as np
import pandas as pd
from IPython.display import display
from sklearn.metrics import (ConfusionMatrixDisplay, accuracy_score, classification_report, f1_score,
                             mean_absolute_error, mean_squared_error, r2_score)

from ml_utils import *          # paths, FEATURES, build_pipeline, split_data, register_model ...
```

```python
df = load_clean_data()
X_train, X_test, yr_train, yr_test, yc_train, yc_test = split_data(df)
print("Training novels:", len(X_train), "| Test novels:", len(X_test))
print("Features:", FEATURES)
```

```python
from sklearn.ensemble import RandomForestRegressor

model = build_pipeline(RandomForestRegressor(n_estimators=100, max_depth=10, n_jobs=-1, random_state=42))
model.fit(X_train, yr_train)

pred = np.clip(model.predict(X_test), 0, 5)          # ratings only exist between 0 and 5
r2 = r2_score(yr_test, pred)
mae = mean_absolute_error(yr_test, pred)
rmse = np.sqrt(mean_squared_error(yr_test, pred))
baseline = mean_absolute_error(yr_test, np.full(len(yr_test), yr_train.mean()))
print(f"R2   : {r2:.4f}   (1 = perfect, 0 = no better than predicting the mean)")
print(f"MAE  : {mae:.4f}   (baseline that always predicts the mean: {baseline:.4f})")
print(f"RMSE : {rmse:.4f}")
```

```python
plt.figure(figsize=(5, 5))
plt.scatter(yr_test, pred, s=8, alpha=0.4, color="#8b3a2f")
plt.plot([1.5, 5], [1.5, 5], "k--", label="perfect prediction")
plt.xlabel("Actual average rating"); plt.ylabel("Predicted average rating")
plt.title("Actual vs predicted (test set)"); plt.legend(); plt.show()
```

```python
names = model.named_steps["prep"].get_feature_names_out()
importance = pd.Series(model.named_steps["model"].feature_importances_, index=names).sort_values().tail(10)
importance.plot.barh(figsize=(7, 4), color="#4a5a6a", title="Ten most important features")
plt.tight_layout(); plt.show()
```

```python
register_model("regression", "random_forest", "Random Forest Regressor", model,
               {"r2": r2, "mae": mae, "rmse": rmse})
```

```python
from sklearn.ensemble import RandomForestClassifier

model = build_pipeline(RandomForestClassifier(n_estimators=100, max_depth=10, n_jobs=-1, random_state=42))
model.fit(X_train, yc_train)

pred = model.predict(X_test)
accuracy = accuracy_score(yc_test, pred)
f1 = f1_score(yc_test, pred, average="weighted")
print(f"Accuracy      : {accuracy:.4f}   (random guessing is about 0.33 with 3 balanced classes)")
print(f"F1 (weighted) : {f1:.4f}")
print()
print(classification_report(yc_test, pred, labels=CLASS_LABELS, zero_division=0))
```

```python
ConfusionMatrixDisplay.from_predictions(yc_test, pred, labels=CLASS_LABELS, cmap="Blues")
plt.title("Confusion matrix (test set)"); plt.show()
```

```python
names = model.named_steps["prep"].get_feature_names_out()
importance = pd.Series(model.named_steps["model"].feature_importances_, index=names).sort_values().tail(10)
importance.plot.barh(figsize=(7, 4), color="#4a5a6a", title="Ten most important features")
plt.tight_layout(); plt.show()
```
