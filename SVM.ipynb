```python
import joblib
import matplotlib.pyplot as plt
import numpy as np
import pandas as pd
from IPython.display import display
from sklearn.metrics import (ConfusionMatrixDisplay, accuracy_score, classification_report, f1_score,
                             mean_absolute_error, mean_squared_error, r2_score)

from ml_utils import *          # paths, FEATURES, build_pipeline, split_data, register_model ...
```

```python
df = load_clean_data()
X_train, X_test, yr_train, yr_test, yc_train, yc_test = split_data(df)
print("Training novels:", len(X_train), "| Test novels:", len(X_test))
print("Features:", FEATURES)
```

```python
from sklearn.svm import SVC

model = build_pipeline(SVC())
model.fit(X_train, yc_train)

pred = model.predict(X_test)
accuracy = accuracy_score(yc_test, pred)
f1 = f1_score(yc_test, pred, average="weighted")
print(f"Accuracy      : {accuracy:.4f}   (random guessing is about 0.33 with 3 balanced classes)")
print(f"F1 (weighted) : {f1:.4f}")
print()
print(classification_report(yc_test, pred, labels=CLASS_LABELS, zero_division=0))
```

```python
ConfusionMatrixDisplay.from_predictions(yc_test, pred, labels=CLASS_LABELS, cmap="Blues")
plt.title("Confusion matrix (test set)"); plt.show()
```

```python
register_model("classification", "svm", "Support Vector Machine (SVM)", model,
               {"accuracy": accuracy, "f1": f1})
```
